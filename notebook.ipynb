```python
# 2. HELPER FUNCTIONS
def get_uploads_playlist_id(channel_id):
    try:
        res = client_youtube.channels().list(part="contentDetails", id=channel_id).execute()
        return res["items"][0]["contentDetails"]["relatedPlaylists"]["uploads"]
    except:
        return None

def get_video_metadata(video_id_list):
    """Fetches stats and duration for a batch of videos"""
    res = client_youtube.videos().list(
        part="snippet,contentDetails,statistics",
        id=",".join(video_id_list)
    ).execute()

    batch_data = []
    for item in res.get("items", []):
        duration = isodate.parse_duration(item["contentDetails"]["duration"]).total_seconds()
        batch_data.append({
            "video_id": item["id"],
            "title": item["snippet"]["title"],
            "description": item["snippet"]["description"],
            "published_at": item["snippet"]["publishedAt"],
            "view_count": int(item["statistics"].get("viewCount", 0)),
            "comment_count": int(item["statistics"].get("commentCount", 0)),
            "duration_sec": duration
        })
    return batch_data

def is_english(text):
    try: return detect(text) == "en"
    except: return False
```

```python
# 3. MAIN DATA COLLECTION LOOP
all_extracted_data = []

for brand, c_id in BRANDS.items():
    print(f"Processing: {brand}...")
    playlist_id = get_uploads_playlist_id(c_id)
    if not playlist_id: 
        print(f"For {brand} I was not able to find anything")
        continue
    else:
        print(f"I found something for {brand}")
    # Get video IDs from uploads
    v_ids = []
    req = client_youtube.playlistItems().list(part="snippet", playlistId=playlist_id, maxResults=50)
    res = req.execute()
    for item in res.get("items", []):
        v_ids.append(item["snippet"]["resourceId"]["videoId"])

    # Get full metadata
    metadata = get_video_metadata(v_ids)
    for video in metadata:
        video["brand"] = brand
        all_extracted_data.append(video)
```

```python
# 1. Load Pre-trained ResNet50
# We use 'pretrained=True' to get the weights learned on ImageNet
resnet = models.resnet50(pretrained=True)
# Remove the last layer (the classifier) to get features instead of labels
resnet = torch.nn.Sequential(*(list(resnet.children())[:-1]))
resnet.eval() # Set to evaluation mode

# 2. Define Image Transformation
preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

def extract_resnet_features(video_path):
    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)

    if fps <= 0: fps = 30

    frame_features = []

    # Primary Loop: Extract 1 frame per second
    count = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        if count % int(fps) == 0:
            img = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            pil_img = Image.fromarray(img)
            input_tensor = preprocess(pil_img).unsqueeze(0)

            with torch.no_grad():
                feature = resnet(input_tensor)
            frame_features.append(feature.flatten().numpy())
        count += 1

    # FALLBACK: If we didn't get any frames, grab the MIDDLE frame
    if not frame_features and total_frames > 0:
        middle_frame_idx = total_frames // 2
        cap.set(cv2.CAP_PROP_POS_FRAMES, middle_frame_idx)
        ret, frame = cap.read()
        if ret:
            img = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            pil_img = Image.fromarray(img)
            input_tensor = preprocess(pil_img).unsqueeze(0)
            with torch.no_grad():
                feature = resnet(input_tensor)
            frame_features.append(feature.flatten().numpy())

    cap.release()

    if not frame_features:
        return np.zeros(2048) # Final safety net

    return np.mean(frame_features, axis=0)

# 3. Process all videos and save features
visual_features_dict = {}
video_files = [f for f in os.listdir(f"{PROJECT_PATH}/videos") if f.endswith('.mp4')]

print("Extracting ResNet features (Visual)...")
for v_file in tqdm(video_files):
    v_id = v_file.replace(".mp4", "")
    v_path = f"{PROJECT_PATH}/videos/{v_file}"

    try:
        features = extract_resnet_features(v_path)
        visual_features_dict[v_id] = features
    except Exception as e:
        print(f"Error processing {v_id}: {e}")

# Save the visual features to a file so you don't have to run this again
np.save(f"{PROJECT_PATH}/visual_features.npy", visual_features_dict)
print("Visual features saved!")
```

```python
def extract_mfcc_features(audio_path):
    # Load audio file (16kHz was set in our ffmpeg step)
    y, sr = librosa.load(audio_path, sr=16000)

    # Extract MFCCs (typically 13 or 20 coefficients)
    # n_mfcc=20 is standard for sentiment/speech tasks
    mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)

    # Aggregate over time: take the mean and standard deviation
    # This captures both the average tone and the "energy" (variance)
    mfccs_mean = np.mean(mfccs, axis=1)
    mfccs_std = np.std(mfccs, axis=1)

    # Combine them into a 40-dimensional vector
    return np.hstack((mfccs_mean, mfccs_std))

# 3. Process all audio files
audio_features_dict = {}
audio_files = [f for f in os.listdir(f"{PROJECT_PATH}/audio") if f.endswith('.wav')]

print("Extracting MFCC features (Audio)...")
for a_file in tqdm(audio_files):
    v_id = a_file.replace(".wav", "")
    a_path = f"{PROJECT_PATH}/audio/{a_file}"

    try:
        features = extract_mfcc_features(a_path)
        audio_features_dict[v_id] = features
    except Exception as e:
        print(f"Error processing {v_id}: {e}")

# Save the audio features
np.save(f"{PROJECT_PATH}/audio_features.npy", audio_features_dict)
print("Audio features saved!")
```
