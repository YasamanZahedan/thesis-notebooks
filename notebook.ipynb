```python

from torchvision.models import resnet50, ResNet50_Weights


VIDEO_DIR = os.path.join(PROJECT_PATH, "videos")
FEATURES_VIDEO_DIR = os.path.join(PROJECT_PATH, "features_resnet")

# Ensure the output directory exists
os.makedirs(FEATURES_VIDEO_DIR, exist_ok=True)

# =========================
# LOAD RESNET50
# =========================
# Check for GPU (CUDA), Apple Silicon (MPS), or fallback to CPU
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps") # For Mac users
else:
    device = torch.device("cpu")

print(f"Using device: {device}")

# Load pre-trained weights and the model
weights = ResNet50_Weights.DEFAULT
model = resnet50(weights=weights).to(device)

# Remove the classification head (leaving the 2048-D feature pooler)
feature_extractor = torch.nn.Sequential(*(list(model.children())[:-1]))
feature_extractor.eval()

# Image transformations (resize, crop, normalize)
preprocess = weights.transforms()

# =========================
# EXTRACTION FUNCTION
# =========================
def extract_video_features(video_path, max_frames=30):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        return None

    fps = cap.get(cv2.CAP_PROP_FPS)
    if fps == 0 or np.isnan(fps):
        fps = 30

    frame_interval = int(fps) # Extract 1 frame per second
    frame_features = []
    frame_count = 0
    success = True

    while success and len(frame_features) < max_frames:
        success, frame = cap.read()

        if success and frame_count % frame_interval == 0:
            # OpenCV loads as BGR, convert to RGB
            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            
            # Convert to PyTorch tensor (C, H, W)
            tensor_frame = torch.from_numpy(frame_rgb).permute(2, 0, 1).float() / 255.0
            input_tensor = preprocess(tensor_frame).unsqueeze(0).to(device)

            # Extract features
            with torch.no_grad():
                features = feature_extractor(input_tensor).squeeze()
                frame_features.append(features.cpu().numpy())

        frame_count += 1

    cap.release()

    if not frame_features:
        return np.zeros(2048)

    # Average the frame features into a single video vector
    video_vector = np.mean(frame_features, axis=0)
    return video_vector

# =========================
# MAIN PROCESSING LOOP
# =========================
print("Starting ResNet50 visual feature extraction...")

# Get list of downloaded mp4 files
video_files = [f for f in os.listdir(VIDEO_DIR) if f.endswith('.mp4')]
successful_extractions = 0

for file_name in tqdm(video_files):
    video_id = file_name.replace('.mp4', '')
    video_path = os.path.join(VIDEO_DIR, file_name)
    feature_path = os.path.join(FEATURES_VIDEO_DIR, f"{video_id}.npy")
    
    # Skip if already extracted
    if os.path.exists(feature_path):
        successful_extractions += 1
        continue
        
    try:
        video_vector = extract_video_features(video_path, max_frames=30)
        
        if video_vector is not None and not np.all(video_vector == 0):
            np.save(feature_path, video_vector)
            successful_extractions += 1
        else:
            print(f"\nFailed to extract features for {video_id} (Corrupted/Empty video)")
    except Exception as e:
        print(f"\nError processing {video_id}: {e}")

# =========================
# VERIFICATION / COUNTING
# =========================
print("\n" + "="*50)
print(f"Total videos in folder: {len(video_files)}")
print(f"Successfully extracted visual vectors: {successful_extractions}")
print("="*50)
```

```python
import warnings

# Ignore librosa warnings
warnings.filterwarnings('ignore')

# =========================
# SETUP PATHS
# =========================
# FIX: Set this to match your actual master folder name!
PROJECT_PATH = "Thesis_Data" 

AUDIO_DIR = os.path.join(PROJECT_PATH, "audio")
FEATURES_AUDIO_DIR = os.path.join(PROJECT_PATH, "features_mfcc")

# Ensure output directory exists
os.makedirs(FEATURES_AUDIO_DIR, exist_ok=True)

# =========================
# MFCC EXTRACTION FUNCTION
# =========================
def extract_mfcc(audio_path, n_mfcc=40):
    try:
        # Load audio at 16kHz
        y, sr = librosa.load(audio_path, sr=16000)
        
        # Compute MFCCs
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
        
        # Average across the time axis (axis=1)
        mfccs_mean = np.mean(mfccs, axis=1)
        
        return mfccs_mean
    except Exception as e:
        print(f"Error processing {audio_path}: {e}")
        return None

# =========================
# MAIN PROCESSING LOOP
# =========================
print("Starting Audio MFCC feature extraction...")

# Get all wav files (Now looking in Thesis_Data/audio)
audio_files = [f for f in os.listdir(AUDIO_DIR) if f.endswith('.wav')]
successful_audio = 0

for file_name in tqdm(audio_files):
    video_id = file_name.replace('.wav', '')
    audio_path = os.path.join(AUDIO_DIR, file_name)
    feature_path = os.path.join(FEATURES_AUDIO_DIR, f"{video_id}.npy")
    
    # Skip if already extracted
    if os.path.exists(feature_path):
        successful_audio += 1
        continue
        
    # Extract features
    audio_vector = extract_mfcc(audio_path, n_mfcc=40)
    
    if audio_vector is not None and not np.all(audio_vector == 0):
        # Save the 40-dimensional array
        np.save(feature_path, audio_vector)
        successful_audio += 1
    else:
        print(f"\nFailed to extract audio features for {video_id}")

# =========================
# VERIFICATION / COUNTING
# =========================
print("\n" + "="*50)
print(f"Total audio files in folder: {len(audio_files)}")
print(f"Successfully extracted MFCC vectors: {successful_audio}")
print("="*50)
```

```python
# =========================
# INSTALL & IMPORT VADER
# =========================
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import pandas as pd
import json
import os
from tqdm import tqdm

nltk.download('vader_lexicon')
sid = SentimentIntensityAnalyzer()

# =========================
# STEP 1: CALCULATE RAW SCORES ONLY ONCE
# =========================
print("Calculating raw VADER compound scores...")
raw_vader_data = []

for index, row in tqdm(df.iterrows(), total=df.shape[0]):
    video_id = row['video_id']
    file_path = f"{PROJECT_PATH}/comments/{video_id}.json"
    
    if os.path.exists(file_path):
        with open(file_path, "r") as f:
            comments = json.load(f)
            
        valid_comments = [c for c in comments if isinstance(c, str) and len(c.strip()) > 2]
        if not valid_comments:
            continue
            
        compound_scores = [sid.polarity_scores(c)['compound'] for c in valid_comments]
        avg_compound = sum(compound_scores) / len(compound_scores)
        
        raw_vader_data.append({
            "video_id": video_id,
            "vader_compound_score": avg_compound
        })

df_vader_raw = pd.DataFrame(raw_vader_data)

# =========================
# STEP 2: TEST MULTIPLE THRESHOLDS
# =========================
# Pair your positive and negative thresholds together
threshold_pairs = [
    (0.05, -0.05),
    (0.10, -0.10),
    (0.19, 0.075),
    (0.2, 0.07)
]

# We will store the results of all threshold tests here
threshold_results = []

for pos_thresh, neg_thresh in threshold_pairs:
    
    def assign_label(score):
        if score >= pos_thresh: return 'pos'
        elif score <= neg_thresh: return 'neg'
        else: return 'neu'
        
    # Apply the thresholds
    temp_labels = df_vader_raw['vader_compound_score'].apply(assign_label)
    
    # Calculate how many videos fell into each category
    distribution = temp_labels.value_counts(normalize=True) * 100
    
    threshold_results.append({
        "Threshold (Pos, Neg)": f"({pos_thresh}, {neg_thresh})",
        "Positive %": distribution.get('pos', 0),
        "Neutral %": distribution.get('neu', 0),
        "Negative %": distribution.get('neg', 0),
        "Labels": temp_labels # Saving the actual labels for later
    })

# Display the results
df_threshold_comparison = pd.DataFrame(threshold_results)
display(df_threshold_comparison[['Threshold (Pos, Neg)', 'Positive %', 'Neutral %', 'Negative %']])
```

```python

from torch.utils.data import Dataset, DataLoader, Subset
from sklearn.model_selection import StratifiedKFold

# Force 100% reproducibility
SEED = 0
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)
    # Force GPU to use deterministic algorithms
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")
print(f"Training on device: {device}\n")

# =========================
# 2. DATA SANITIZATION (The Intersection)
# =========================
PROJECT_PATH = "Thesis_Data"
FEATURES_VIDEO_DIR = os.path.join(PROJECT_PATH, "features_resnet")
FEATURES_AUDIO_DIR = os.path.join(PROJECT_PATH, "features_mfcc")

# Find videos that actually exist on your hard drive
video_files = set([f.replace('.npy', '') for f in os.listdir(FEATURES_VIDEO_DIR) if f.endswith('.npy')])
audio_files = set([f.replace('.npy', '') for f in os.listdir(FEATURES_AUDIO_DIR) if f.endswith('.npy')])

# Intersection: Must have video AND audio AND be in your VADER dataframe
ready_videos = video_files.intersection(audio_files)
df_nn_ready = df_vader[df_vader['video_id'].isin(ready_videos)].reset_index(drop=True)

print(f"Final safe dataset size: {len(df_nn_ready)} videos ready for Fusion.\n")

# =========================
# 3. DATASET LOADER
# =========================
class ThesisMultimodalDataset(Dataset):
    def __init__(self, dataframe, visual_dir, audio_dir):
        self.data = dataframe
        self.visual_dir = visual_dir
        self.audio_dir = audio_dir
        self.label_map = {'neg': 0, 'neu': 1, 'pos': 2}

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        row = self.data.iloc[idx]
        video_id = row['video_id']
        
        v_path = os.path.join(self.visual_dir, f"{video_id}.npy")
        a_path = os.path.join(self.audio_dir, f"{video_id}.npy")
        
        visual_feature = torch.tensor(np.load(v_path), dtype=torch.float32)
        audio_feature = torch.tensor(np.load(a_path), dtype=torch.float32)
        
        label = torch.tensor(self.label_map[row['vader_label']], dtype=torch.long)
        return visual_feature, audio_feature, label

full_dataset = ThesisMultimodalDataset(df_nn_ready, FEATURES_VIDEO_DIR, FEATURES_AUDIO_DIR)
all_labels = [full_dataset.label_map[label] for label in df_nn_ready['vader_label']]

# =========================
# 4. MULTIMODAL FUSION NETWORK
# =========================
class MultimodalSentimentNN(nn.Module):
    def __init__(self):
        super(MultimodalSentimentNN, self).__init__()
        
        # Branch 1: Compress the massive 2048-D visual vector down to 256
        self.visual_branch = nn.Sequential(
            nn.Linear(2048, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.5)
        )
        
        # Branch 2: Upscale the tiny 40-D audio vector to 128
        self.audio_branch = nn.Sequential(
            nn.Linear(40, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.3)
        )
        
        # Branch 3: The Fusion Head (256 + 128 = 384 dimensions)
        self.classifier = nn.Sequential(
            nn.Linear(384, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(128, 3) # Output: 3 neurons for Neg, Neu, Pos
        )

    def forward(self, visual_x, audio_x):
        v_out = self.visual_branch(visual_x)
        a_out = self.audio_branch(audio_x)
        
        # The actual "Fusion" happens here: merging the two arrays side-by-side
        fused = torch.cat((v_out, a_out), dim=1)
        
        return self.classifier(fused)

# =========================
# 5. STRATIFIED K-FOLD TRAINING
# =========================
K_FOLDS = 5
EPOCHS = 25
skf = StratifiedKFold(n_splits=K_FOLDS, shuffle=True, random_state=SEED) # Seed locked to 0!

fold_results = []
fold_loss_history = {} 

print(f"Starting {K_FOLDS}-Fold Cross Validation...\n")
dummy_x = np.zeros(len(all_labels)) 

for fold, (train_idx, test_idx) in enumerate(skf.split(dummy_x, all_labels)):
    print(f"--- Training FOLD {fold+1} ---")
    
    train_sub = Subset(full_dataset, train_idx)
    test_sub = Subset(full_dataset, test_idx)
    
    # We set generator with seed to ensure the shuffling inside dataloader is identical every run
    g = torch.Generator()
    g.manual_seed(SEED)
    
    train_loader = DataLoader(train_sub, batch_size=16, shuffle=True, generator=g)
    test_loader = DataLoader(test_sub, batch_size=16, shuffle=False)
    
    # Re-initialize model per fold to prevent data leakage
    model = MultimodalSentimentNN().to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-3)
    
    epoch_losses = []
    
    for epoch in range(EPOCHS):
        model.train()
        running_loss = 0.0
        
        for visuals, audios, labels in train_loader:
            visuals, audios, labels = visuals.to(device), audios.to(device), labels.to(device)
            
            optimizer.zero_grad()
            outputs = model(visuals, audios)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            
        epoch_losses.append(running_loss / len(train_loader))
        
    fold_loss_history[f"Fold {fold+1}"] = epoch_losses
            
    # Evaluate Fold
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for visuals, audios, labels in test_loader:
            visuals, audios, labels = visuals.to(device), audios.to(device), labels.to(device)
            outputs = model(visuals, audios)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
    fold_accuracy = 100.0 * correct / total
    fold_results.append(fold_accuracy)
    print(f"Fold {fold+1} Test Accuracy: {fold_accuracy:.2f}%\n")

avg_acc = np.mean(fold_results)
std_acc = np.std(fold_results)

print("="*40)
print(f"✅ FINAL AVERAGE ACCURACY: {avg_acc:.2f}% (+/- {std_acc:.2f}%)")
print("="*40)

```
