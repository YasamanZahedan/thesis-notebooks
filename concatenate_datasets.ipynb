```python
import numpy as np
import os

def load_and_fix(file_path):
    data = np.load(file_path, allow_pickle=True)
    # If the array is 0-dimensional, it's likely a wrapped object (like a dict)
    if data.ndim == 0:
        print(f"Unpacking 0-dim object from {file_path}...")
        return data.item() 
    return data

# 1. Load the extracted features
features1 = load_and_fix(r'C:\Users\yasam\Desktop\thesis-notebooks\Thesis_Data\visual_features_clip.npy')
features2 = load_and_fix(r'C:\Users\yasam\Desktop\thesis-notebooks\Thesis_Data\visual_features_clip2.npy')

# 2. Check the types
print(f"Type 1: {type(features1)}")
print(f"Type 2: {type(features2)}")

# 3. Handle based on what they actually are
if isinstance(features1, dict) and isinstance(features2, dict):
    # If they are dictionaries, merge them like this:
    merged_features = {**features1, **features2}
    print(f"Merged as dictionary. Total keys: {len(merged_features)}")
    np.save('visual_features_clip_combined.npy', merged_features)
else:
    # If they are standard arrays, concatenate
    merged_features = np.concatenate((features1, features2), axis=0)
    print(f"Merged as array. Final shape: {merged_features.shape}")
    np.save('visual_features_clip_combined.npy', merged_features)
```

```python
# To load the combined file later:
combined_data = np.load('visual_features_clip_combined.npy', allow_pickle=True).item()

# To access features for a specific video:
video_name = "_7CjZh6y4xQ"
if video_name in combined_data:
    vector = combined_data[video_name]
    print(f"Features for {video_name}: {vector.shape}")
```

```python
import numpy as np

def load_and_fix(file_path):
    data = np.load(file_path, allow_pickle=True)
    if data.ndim == 0:
        return data.item() 
    return data

# 1. Paths to your audio features
path1 = r'C:\Users\yasam\Desktop\thesis-notebooks\Thesis_Data\audio_features_vggish.npy'
path2 = r'C:\Users\yasam\Desktop\thesis-notebooks\Thesis_Data\audio_features_vggish2.npy'

# 2. Load the audio dictionaries
audio_features1 = load_and_fix(path1)
audio_features2 = load_and_fix(path2)

print(f"Set 1 audio entries: {len(audio_features1)}")
print(f"Set 2 audio entries: {len(audio_features2)}")

# 3. Merge dictionaries
# This combines both sets. If a key exists in both, set 2 overwrites set 1.
merged_audio = {**audio_features1, **audio_features2}

print(f"Total merged audio entries: {len(merged_audio)}")

# 4. Save the combined audio features
save_path = r'C:\Users\yasam\Desktop\thesis-notebooks\Thesis_Data\audio_features_vggish_combined.npy'
np.save(save_path, merged_audio)

print(f"Successfully saved combined audio to: {save_path}")
```
