```python
# MASTER CELL: Data Loading, Late Fusion MLP, & LOOCV
import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import LeaveOneOut
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, accuracy_score
from tqdm import tqdm  # Added for progress tracking

# ==========================================
# 1. LOAD AND PREPARE DATA
# ==========================================
PROJECT_PATH = "Thesis_Data"
LABELS_PATH = "videos_with_sentiment_labels_combined.csv"

df = pd.read_csv(LABELS_PATH)
visual_dict = np.load(f"{PROJECT_PATH}/visual_features_clip_combined.npy", allow_pickle=True).item()
audio_dict = np.load(f"{PROJECT_PATH}/audio_features_vggish_combined.npy", allow_pickle=True).item()

X_visual, X_audio, y_labels = [], [], []

for index, row in df.iterrows():
    v_id = row['video_id']
    label = row['majority_sentiment']
    if v_id in visual_dict and v_id in audio_dict:
        X_visual.append(visual_dict[v_id])
        X_audio.append(audio_dict[v_id])
        y_labels.append(label)

X_visual = np.array(X_visual)
X_audio = np.array(X_audio)
y_labels = np.array(y_labels)

le = LabelEncoder()
y_encoded = le.fit_transform(y_labels)

# Calculate class weights for the imbalanced dataset
weights = compute_class_weight(class_weight='balanced', classes=np.unique(y_encoded), y=y_encoded)
class_weights_tensor = torch.tensor(weights, dtype=torch.float32)

# ==========================================
# 2. DEFINE THE 2-LAYER LATE FUSION MLP
# ==========================================
class SimpleLateFusionMLP(nn.Module):
    def __init__(self, visual_dim, audio_dim, num_classes=3):
        super(SimpleLateFusionMLP, self).__init__()
        
        # Late Fusion: we concatenate immediately, so input is the sum of both dims
        input_dim = visual_dim + audio_dim
        
        self.network = nn.Sequential(
            nn.Linear(input_dim, 256),    # Layer 1
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.5),              # Dropout requested by user
            nn.Linear(256, num_classes)   # Layer 2 (Output)
        )

    def forward(self, visual_x, audio_x):
        # Concatenate modalities along the feature dimension
        combined = torch.cat((visual_x, audio_x), dim=1) 
        return self.network(combined)

# ==========================================
# 3. SETUP LOOCV TRAINING
# ==========================================
loo = LeaveOneOut()

all_true_labels = []
all_predictions = []

# Fixed epochs because Early Stopping doesn't work with a validation set size of 1
FIXED_EPOCHS = 25 
BATCH_SIZE = 32

print(f"\nStarting Leave-One-Out CV ({len(X_visual)} iterations)...")

# Wrap the loo.split generator in tqdm for a progress bar
for train_idx, test_idx in tqdm(loo.split(X_visual), total=len(X_visual), desc="LOOCV Progress"):
    
    # -- Split Data --
    X_v_train, X_v_test = X_visual[train_idx], X_visual[test_idx]
    X_a_train, X_a_test = X_audio[train_idx], X_audio[test_idx]
    y_train, y_test = y_encoded[train_idx], y_encoded[test_idx]
    
    # -- Scale Data (Fit on Train, Transform on Train & Test) --
    scaler_v = StandardScaler()
    X_v_train_scaled = scaler_v.fit_transform(X_v_train)
    X_v_test_scaled = scaler_v.transform(X_v_test)
    
    scaler_a = StandardScaler()
    X_a_train_scaled = scaler_a.fit_transform(X_a_train)
    X_a_test_scaled = scaler_a.transform(X_a_test)
    
    # -- Convert to Tensors --
    train_v = torch.tensor(X_v_train_scaled, dtype=torch.float32)
    train_a = torch.tensor(X_a_train_scaled, dtype=torch.float32)
    train_y = torch.tensor(y_train, dtype=torch.long)
    
    test_v = torch.tensor(X_v_test_scaled, dtype=torch.float32)
    test_a = torch.tensor(X_a_test_scaled, dtype=torch.float32)
    
    # Create DataLoader for training
    train_dataset = torch.utils.data.TensorDataset(train_v, train_a, train_y)
    train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
    
    # -- Initialize Model, Loss, Optimizer --
    model = SimpleLateFusionMLP(visual_dim=X_visual.shape[1], audio_dim=X_audio.shape[1], num_classes=len(le.classes_))
    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor) 
    optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-4)
    
    # -- TRAINING PHASE --
    model.train()
    for epoch in range(FIXED_EPOCHS):
        for batch_v, batch_a, batch_y in train_loader:
            optimizer.zero_grad()
            outputs = model(batch_v, batch_a) 
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()
            
    # -- TESTING PHASE (On the 1 left-out sample) --
    model.eval()
    with torch.no_grad():
        test_out = model(test_v, test_a)
        _, pred = torch.max(test_out, 1)
        
        all_predictions.append(pred.item())
        all_true_labels.append(y_test[0])

# ==========================================
# 4. FINAL EVALUATION
# ==========================================
print("\n" + "="*45)
print("=== FINAL LOOCV LATE FUSION RESULTS ===")
print("="*45)
print(f"Overall Accuracy: {accuracy_score(all_true_labels, all_predictions):.4f}\n")
print("Detailed Classification Report:")
print(classification_report(all_true_labels, all_predictions, target_names=le.classes_))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score
import numpy as np
import copy  # Needed to save the best model weights

# Assuming X_visual, X_audio, y_encoded, le, and class_weights_tensor 
# are already defined from your data loading cell.

# ==========================================
# SETUP STRATIFIED K-FOLD & EARLY STOPPING
# ==========================================
k_folds = 10
skf = StratifiedKFold(n_splits=k_folds, shuffle=True, random_state=42)

all_true_labels = []
all_predictions = []
fold_accuracies = []

MAX_EPOCHS = 100
PATIENCE = 15  # Stop if validation loss doesn't improve for 15 epochs
BATCH_SIZE = 32

print(f"\nStarting Stratified {k_folds}-Fold CV with Early Stopping...")

for fold, (train_idx, val_idx) in enumerate(skf.split(X_visual, y_encoded)):
    print(f"\n--- FOLD {fold + 1}/{k_folds} ---")
    
    # -- 1. Split Data --
    X_v_train, X_v_val = X_visual[train_idx], X_visual[val_idx]
    X_a_train, X_a_val = X_audio[train_idx], X_audio[val_idx]
    y_train, y_val = y_encoded[train_idx], y_encoded[val_idx]
    
    # -- 2. Scale Data (Fit on Train, Transform on Val) --
    scaler_v = StandardScaler()
    X_v_train_scaled = scaler_v.fit_transform(X_v_train)
    X_v_val_scaled = scaler_v.transform(X_v_val)
    
    scaler_a = StandardScaler()
    X_a_train_scaled = scaler_a.fit_transform(X_a_train)
    X_a_val_scaled = scaler_a.transform(X_a_val)
    
    # -- 3. Create DataLoaders --
    train_loader = torch.utils.data.DataLoader(
        torch.utils.data.TensorDataset(
            torch.tensor(X_v_train_scaled, dtype=torch.float32), 
            torch.tensor(X_a_train_scaled, dtype=torch.float32),
            torch.tensor(y_train, dtype=torch.long)
        ), batch_size=BATCH_SIZE, shuffle=True
    )
    
    val_loader = torch.utils.data.DataLoader(
        torch.utils.data.TensorDataset(
            torch.tensor(X_v_val_scaled, dtype=torch.float32), 
            torch.tensor(X_a_val_scaled, dtype=torch.float32),
            torch.tensor(y_val, dtype=torch.long)
        ), batch_size=BATCH_SIZE, shuffle=False
    )
    
    # -- 4. Initialize Model, Loss, Optimizer --
    # (Assuming SimpleLateFusionMLP is defined from the previous step)
    model = SimpleLateFusionMLP(visual_dim=X_visual.shape[1], audio_dim=X_audio.shape[1], num_classes=len(le.classes_))
    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor) 
    optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-4)
    
    # -- 5. Training Loop with Early Stopping --
    best_val_loss = float('inf')
    epochs_no_improve = 0
    best_model_weights = None
    
    for epoch in range(MAX_EPOCHS):
        # Training Phase
        model.train()
        for batch_v, batch_a, batch_y in train_loader:
            optimizer.zero_grad()
            outputs = model(batch_v, batch_a) 
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()
            
        # Validation Phase
        model.eval()
        val_loss = 0.0
        with torch.no_grad():
            for batch_v, batch_a, batch_y in val_loader:
                outputs = model(batch_v, batch_a)
                loss = criterion(outputs, batch_y)
                val_loss += loss.item()
                
        val_loss /= len(val_loader)
        
        # Early Stopping Check
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            epochs_no_improve = 0
            # Save the best weights! Don't just save the model at the last overfitted epoch.
            best_model_weights = copy.deepcopy(model.state_dict())
        else:
            epochs_no_improve += 1
            
        if epochs_no_improve >= PATIENCE:
            print(f"  Early stopping at Epoch {epoch + 1} (Best Val Loss: {best_val_loss:.4f})")
            break
            
    # -- 6. Evaluate Fold Using BEST Weights --
    # Restore the model to its best state before making the final fold predictions
    if best_model_weights is not None:
        model.load_state_dict(best_model_weights)
        
    model.eval()
    fold_preds, fold_labels = [], []
    with torch.no_grad():
        for batch_v, batch_a, batch_y in val_loader:
            outputs = model(batch_v, batch_a)
            _, preds = torch.max(outputs, 1)
            fold_preds.extend(preds.numpy())
            fold_labels.extend(batch_y.numpy())
            
    fold_acc = accuracy_score(fold_labels, fold_preds)
    fold_accuracies.append(fold_acc)
    print(f"  Fold Accuracy: {fold_acc:.4f}")
    
    # Store predictions for the final global report
    all_true_labels.extend(fold_labels)
    all_predictions.extend(fold_preds)

# ==========================================
# FINAL EVALUATION
# ==========================================
print("\n" + "="*45)
print("=== FINAL 10-FOLD LATE FUSION RESULTS ===")
print("="*45)
print(f"Average Accuracy across folds: {np.mean(fold_accuracies):.4f} (+/- {np.std(fold_accuracies):.4f})\n")
print("Detailed Classification Report (Aggregated):")
print(classification_report(all_true_labels, all_predictions, target_names=le.classes_))
```

### Custom Validation: Stratified "Leave-Three-Out"
Standard $K$-Fold or random splitting might accidentally create test sets with *zero* negative videos. To guarantee a rigorous evaluation, we use a custom generator:
* In every iteration, the test set is forced to contain **exactly 1 Positive, 1 Neutral, and 1 Negative video**.
* The remaining videos form the training set.
* The loop runs exactly $N$ times, where $N$ is the size of the smallest class (e.g., 69 iterations), ensuring every single minority sample is tested exactly once.

### Cost-Sensitive Learning (Weighted Loss)
To prevent the model from artificially boosting its accuracy by simply ignoring the minority class, we compute balancing weights using Scikit-Learn. These weights are passed directly into the PyTorch `CrossEntropyLoss` function, heavily penalizing the model for misclassifying the rare Negative samples.

### Why this is Intermediate Fusion
look at forward method, the fusion happens halfway through the data's journey:

Parallel Processing: The Visual and Audio data enter their own "branches" (visual_branch and audio_branch) to be compressed into abstract features.

The "Handshake": used torch.cat((v_features, a_features), dim=1) to physically glue these feature vectors together into one long vector.

Joint Learning: This combined vector is then passed through a shared classifier. This allows the model to learn relationships between the modalities (e.g., "if the person is smiling AND the audio pitch is high, it's Positive").

```python
import os
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, accuracy_score

X_visual, X_audio, y_labels = [], [], []

for index, row in df.iterrows():
    v_id = row['video_id']
    label = row['majority_sentiment']
    if v_id in visual_dict and v_id in audio_dict:
        X_visual.append(visual_dict[v_id])
        X_audio.append(audio_dict[v_id])
        y_labels.append(label)

X_visual = np.array(X_visual)
X_audio = np.array(X_audio)
y_labels = np.array(y_labels)

le = LabelEncoder()
y_encoded = le.fit_transform(y_labels)

# Calculate class weights for the imbalanced dataset
weights = compute_class_weight(class_weight='balanced', classes=np.unique(y_encoded), y=y_encoded)
class_weights_tensor = torch.tensor(weights, dtype=torch.float32)
# ==========================================
# 2. DEFINE THE MODEL (Moved OUTSIDE the loop)
# ==========================================
class IntermediateFusionMLP(nn.Module):
    def __init__(self, visual_dim, audio_dim, num_classes=3):
        super(IntermediateFusionMLP, self).__init__()
        
        # 1. VISUAL BRANCH: Compress and process CLIP features
        self.visual_branch = nn.Sequential(
            nn.Linear(visual_dim, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.4)
        )
        
        # 2. AUDIO BRANCH: Compress and process VGGish features
        self.audio_branch = nn.Sequential(
            nn.Linear(audio_dim, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.4)
        )
        
        # 3. FUSION & CLASSIFICATION (128 + 64 = 192)
        self.classifier = nn.Sequential(
            nn.Linear(192, 64), 
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, num_classes)
        )

    def forward(self, visual_x, audio_x):
        v_features = self.visual_branch(visual_x)
        a_features = self.audio_branch(audio_x)
        merged_vector = torch.cat((v_features, a_features), dim=1)
        return self.classifier(merged_vector)

# ==========================================
# 3. DEFINE CUSTOM GENERATOR
# ==========================================
def stratified_leave_three_out(y):
    """Custom generator to yield exactly 1 sample per class for the test set."""
    idx_class_0 = np.where(y == 0)[0].tolist() 
    idx_class_1 = np.where(y == 1)[0].tolist() 
    idx_class_2 = np.where(y == 2)[0].tolist() 
    
    random.shuffle(idx_class_0)
    random.shuffle(idx_class_1)
    random.shuffle(idx_class_2)
    
    min_class_count = min(len(idx_class_0), len(idx_class_1), len(idx_class_2))
    
    for i in range(min_class_count):
        test_idx = [idx_class_0[i], idx_class_1[i], idx_class_2[i]]
        train_idx = list(set(range(len(y))) - set(test_idx))
        yield train_idx, test_idx

# ==========================================
# 4. START EVALUATION LOOP
# ==========================================
all_true_labels = []
all_predictions = []

FIXED_EPOCHS = 25 
BATCH_SIZE = 32

print(f"\nStarting Stratified Leave-One-Per-Class-Out CV...")

for i, (train_idx, test_idx) in enumerate(stratified_leave_three_out(y_encoded)):
    
    # -- 1. Split Data --
    X_v_train, X_v_test = X_visual[train_idx], X_visual[test_idx]
    X_a_train, X_a_test = X_audio[train_idx], X_audio[test_idx]
    y_train, y_test = y_encoded[train_idx], y_encoded[test_idx]
    
    # -- 2. Scale Data --
    scaler_v = StandardScaler()
    X_v_train_scaled = scaler_v.fit_transform(X_v_train)
    X_v_test_scaled = scaler_v.transform(X_v_test)
    
    scaler_a = StandardScaler()
    X_a_train_scaled = scaler_a.fit_transform(X_a_train)
    X_a_test_scaled = scaler_a.transform(X_a_test)
    
    # -- 3. Convert to Tensors --
    train_v = torch.tensor(X_v_train_scaled, dtype=torch.float32)
    train_a = torch.tensor(X_a_train_scaled, dtype=torch.float32)
    train_y = torch.tensor(y_train, dtype=torch.long)
    
    test_v = torch.tensor(X_v_test_scaled, dtype=torch.float32)
    test_a = torch.tensor(X_a_test_scaled, dtype=torch.float32)
    
    # -- 4. Create DataLoader --
    train_dataset = TensorDataset(train_v, train_a, train_y)
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
    
    # -- 5. Initialize Model, Loss, Optimizer --
    # THIS IS WHAT WAS MISSING: You must instantiate the model here!
    model = IntermediateFusionMLP(visual_dim=X_visual.shape[1], audio_dim=X_audio.shape[1], num_classes=len(le.classes_))
    
    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor) 
    optimizer = optim.Adam(model.parameters(), lr=0.0001, weight_decay=1e-4)
    
    # -- 6. TRAINING PHASE --
    model.train()
    for epoch in range(FIXED_EPOCHS):
        for batch_v, batch_a, batch_y in train_loader:
            optimizer.zero_grad()
            outputs = model(batch_v, batch_a) 
            loss = criterion(outputs, batch_y)
            loss.backward()
            optimizer.step()
            
    # -- 7. TESTING PHASE --
    model.eval()
    with torch.no_grad():
        test_out = model(test_v, test_a)
        _, preds = torch.max(test_out, 1)
        
        all_predictions.extend(preds.tolist())
        all_true_labels.extend(y_test.tolist())
        
    if (i + 1) % 10 == 0:
        # Calculate exactly how many iterations based on minority class
        total_iters = min(np.bincount(y_encoded))
        print(f"  Completed iteration {i + 1} / {total_iters}...")

# ==========================================
# 8. FINAL EVALUATION
# ==========================================
print("\n" + "="*45)
print("=== FINAL LEAVE-THREE-OUT RESULTS ===")
print("="*45)
print(f"Overall Accuracy: {accuracy_score(all_true_labels, all_predictions):.4f}\n")
print("Detailed Classification Report:")
print(classification_report(all_true_labels, all_predictions, target_names=le.classes_))
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import LeaveOneOut
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import classification_report, accuracy_score
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from tqdm import tqdm

# ==========================================
# 1. LOAD AND PREPARE DATA
# ==========================================
PROJECT_PATH = "Thesis_Data"
LABELS_PATH = "videos_with_sentiment_labels_combined.csv"

df = pd.read_csv(LABELS_PATH)
visual_dict = np.load(f"{PROJECT_PATH}/visual_features_clip_combined.npy", allow_pickle=True).item()
audio_dict = np.load(f"{PROJECT_PATH}/audio_features_vggish_combined.npy", allow_pickle=True).item()

X_visual, X_audio, y_labels = [], [], []

for index, row in df.iterrows():
    v_id = row['video_id']
    label = row['majority_sentiment']
    if v_id in visual_dict and v_id in audio_dict:
        X_visual.append(visual_dict[v_id])
        X_audio.append(audio_dict[v_id])
        y_labels.append(label)

X_visual = np.array(X_visual)
X_audio = np.array(X_audio)
y_labels = np.array(y_labels)

# Encode labels
le = LabelEncoder()
y_encoded = le.fit_transform(y_labels)

# ==========================================
# 2. LATE FUSION (Concatenation)
# ==========================================
# Stack features horizontally: e.g., 512 (CLIP) + 128 (VGGish) = 640 features
X_combined = np.hstack((X_visual, X_audio))
print(f"Combined Feature Shape: {X_combined.shape}")

# ==========================================
# 3. SELECT MODEL
# ==========================================
# Change to "RF" if you want to try Random Forest instead
CHOSEN_MODEL = "XGB" 

if CHOSEN_MODEL == "XGB":
    # XGBoost is great for dense embeddings. Depth is kept shallow (3) to prevent overfitting.
    model = XGBClassifier(
        n_estimators=100, 
        max_depth=3, 
        learning_rate=0.1, 
        random_state=42, 
        n_jobs=-1
    )
else:
    # Random Forest baseline
    model = RandomForestClassifier(
        n_estimators=200, 
        max_depth=None, 
        random_state=42, 
        n_jobs=-1
    )

# ==========================================
# 4. LOOCV TRAINING & EVALUATION LOOP
# ==========================================
loo = LeaveOneOut()
all_true_labels = []
all_predictions = []

print(f"\nStarting Leave-One-Out CV with {CHOSEN_MODEL} ({len(X_combined)} iterations)...")

for train_idx, test_idx in tqdm(loo.split(X_combined), total=len(X_combined), desc="LOOCV Progress"):
    
    # -- Split Data --
    X_train, X_test = X_combined[train_idx], X_combined[test_idx]
    y_train, y_test = y_encoded[train_idx], y_encoded[test_idx]
    
    # -- Scale Data --
    # While trees are scale-invariant, it is highly recommended to standardize 
    # when combining embeddings from completely different spaces (CLIP vs VGGish).
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # -- Handle Imbalance Dynamically --
    # Calculate sample weights for the 499 training samples in this specific fold
    sample_weights = compute_sample_weight(class_weight='balanced', y=y_train)
    
    # -- Train & Predict --
    model.fit(X_train_scaled, y_train, sample_weight=sample_weights)
    pred = model.predict(X_test_scaled)
    
    # -- Store Results --
    all_predictions.append(pred[0])
    all_true_labels.append(y_test[0])

# ==========================================
# 5. FINAL EVALUATION
# ==========================================
print("\n" + "="*45)
print(f"=== FINAL LOOCV RESULTS ({CHOSEN_MODEL}) ===")
print("="*45)
print(f"Overall Accuracy: {accuracy_score(all_true_labels, all_predictions):.4f}\n")
print("Detailed Classification Report:")
print(classification_report(all_true_labels, all_predictions, target_names=le.classes_))
```
